```python
print("""
### 1. 고객 등급 분류 (조건문)

고객의 구매 금액에 따라 등급을 나누는 프로그램을 작성하세요.

- 100만 원 이상: **VIP**
- 50만 원 이상: **Gold**
- 10만 원 이상: **Silver**
- 그 외: **Bronze**

👉 입력: 구매 금액
👉 출력: 등급 문자열
""")
```

```python
amount = int(input("구매 금액 입력 : (만원)"))

if amount >= 100 :
    print("고객님의 등급은 VIP 입니다")
elif amount >= 50 :
    print("고객님의 등급은 Gold 입니다")
elif amount >= 10 :
    print("고객님의 등급은 Silver 입니다")
else :
    print("고객님의 등급은 Bronze 입니다")
```

```python
print(
"""
### 3. 재고 소진 알림 (조건문 + 반복문)

상품 재고가 10개 있습니다.

고객이 차례로 주문할 때마다 재고를 차감하고, 재고가 0이 되면 "품절되었습니다"를 출력한 뒤 주문 처리를 중단하세요.

👉 입력: 주문 수량 리스트 `[2, 3, 5, 1]`

👉 출력: 각 주문 처리 결과 + 품절 알림
""")
```

```python
a = list()
b = 10

while True :
    order = input(int("주문 수량을 입력하세요 : "))
    a.append(order)
    b -= order

    if b != 0 :
        print(f"주문 처리 완료. 남은 재고는 {b}개 입니다.")
        continue
    else :
        print("품절입니다")
        break

print(a)

```
