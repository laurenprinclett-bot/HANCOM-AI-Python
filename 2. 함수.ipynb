```python

def mul(*values) :
    base = 1 # 함수 안에서 사용하려면 함수 내에 지역변수로 선언해줘야 함
    
    #print(values)
    #print(type(values))
    for i in range(len(values)) :
        base *= values[i]
        #print(base)
    return base

    #for i in values :
        #base *= i
    #return base
    
```

```python
# 정의부

def comMul(num1, num2) :
    # 1 ~ 100까지 공배수를 구하는 함수 구하기
    # 하나의 케이스로 틀을 잡자 !

    result = list()

    for i in range(1,51) :
        if num1 > num2 :
            if num1 % num2 == 0 :
                choi = num1
            else :
                choi = num1 * num2
        else :
            if num2 % num1 == 0 :
                choi = num2
            else : 
                choi = num1 * num2
        
        if i*choi > 100 :
            break
        else :
            result.append(i*choi)
    return result 

    # 가장 비효율적인 코드 ...
    # 어떤 랜덤 수에 대해 두 수를 나눴을 때 나머지가 모두 0인 경우가 공배수임..
```

```python
input_data = input("두 수 입력 예) 2,3 > ")
input_data
```

```python
num1, num2 = input_data.split(",")
num1, num2 = int(num1), int(num2)
```

```python
comMul(num1, num2)
```
