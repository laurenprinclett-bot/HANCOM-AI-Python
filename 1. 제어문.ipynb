# 반복문 - for

```python
nums = [233, 310, 4, 6, 200, 100, 20, 50, 40, 101]
len(nums)
```

```python
# 100 이상의 숫자만 출력하기
for num in nums :
    if num >= 100 :
        print(f"- 100 이상의 수: {num}")
```

```python
# 100 이상의 숫자, 100 미만인 숫자 각각 리스트에 넣기

a = list()
b = list()

for num in nums :
    if num >= 100 :
        b.append(num)
    else :
        a.append(num)

a.sort()
b.sort()

#print("- 100 미만의 수 리스트 :", a)
print(f"- 100 미만의 수 리스트 : {a}")
print("- 100 이상의 수 리스트 :", b)
```

# for 문 + dictionary 

```python
# nums 의 숫자가 각각 몇 번 반복하는지 dict 구조로 만들어서 출력
```

```python
nums = [1,2,6,7,4,3,2,1,5,9,4,9,7,2,1,3,4,5,6,7,5,6,7,5]

counter = dict()

for num in nums :

    # 키값 판단
    if num not in counter :
    # 그 숫자가 제일 처음 나왔을 때, 키값 + 카운트
        counter[num] = 1

    else : 
    # 그 숫자가 처음 나온 수가 아닌 경우, 카운트
        counter[num] += 1

print(counter)
```
